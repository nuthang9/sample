```python
marks = [90, 25, 67, 45, 80]

num = 1
for score in marks:
    if score >= 60:
        print(num, '번 학생은 합격입니다.')
    else:
        print(num, '번 학생은 불합격입니다.')
    num = num + 1
```

```python
marks = [90, 25, 67, 45, 80]

for i in range(len(marks)):
    if marks[i] >= 60:
        print(i+1, '번 학생은 합격입니다.')
    else:
        print(i+1, '번 학생은 불합격입니다.')
```

```python
print("끝말잇기 게임")
print("시작")
var = "가나다"

print(var)
while True:
    word = input("단어를 입력하세요: ")
    if word[0] == var[-1]:
        var = word
    else:
        print("끝말잇기 실패")
        break
```

```python
print("끝말잇기 게임")
print("시작")
var = "가나다"
used = [var]

print(var)
while True:
    word = input("단어를 입력해주세요 : ")
    print("단어를 입력해주세요 :", word)
    
    if word[0] != var[-1]:
        print("땡")
        break
    
    if word in used:
        print("땡")
        break
    
    used.append(word)
    var = word
```

```python
addr = {
    '박지민': '010-1111-1234',
    '민윤기': '010-2222-1234',
    '김태형': '010-3333-1234'
}

while True:
    menu = input("메뉴 선택(1:추가, 2:검색, 0:종료)? ")
    
    if menu == '1':
        name = input("친구 이름 입력>> ")
        phone = input("전화번호 입력>> ")
        addr[name] = phone
        
    elif menu == '2':
        print("검색가능한 친구들:", list(addr.keys()))
        name = input("친구 이름 입력>> ")
        print(name + "의 연락처 :", addr[name])
        
    elif menu == '0':
        print("프로그램을 종료합니다.")
        break
        
    else:
        print("메뉴 선택 오류입니다. 다시 선택하세요.")
```

```python
morse = {
    '.-':'a','-...':'b','-.-.':'c','-..':'d','.':'e','..-.':'f',
    '--.':'g','....':'h','..':'i','.---':'j','-.-':'k','.-..':'l',
    '--':'m','-.':'n','---':'o','.--.':'p','--.-':'q','.-.':'r',
    '...':'s','-':'t','..-':'u','...-':'v','.--':'w','-..-':'x',
    '-.--':'y','--..':'z'
}

codes = [".... . .-.. .-.. ---", ".--. -.-- - .... --- -."]

for code in codes:
    result = ''
    for c in code.split(' '):
        result = result + morse[c]
    print(result)
```

```python
def num_max(a, b, c):
    if a >= b and a >= c:
        return a
    elif b >= a and b >= c:
        return b
    else:
        return c

num = num_max(10, 12, 7)
num     # 12
```
